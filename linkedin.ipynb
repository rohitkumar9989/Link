```python
import torch
from PIL import Image
import requests
from transformers import AutoProcessor, AutoModelForZeroShotObjectDetection

model_id = "IDEA-Research/grounding-dino-base"

processor = AutoProcessor.from_pretrained(model_id)
model = AutoModelForZeroShotObjectDetection.from_pretrained(model_id)

```

```python
image_path="a.png"

converted_image=Image.open(image_path).convert('RGB')
text_prompt="Apply button . Submitting Resume Button . submit application button . apply now button"
inputs=processor(images=converted_image, text=text_prompt, return_tensors="pt")

with torch.no_grad():
    outputs=model(**inputs)
```

```python
results=processor.post_process_grounded_object_detection(outputs, inputs.input_ids, target_sizes=[converted_image.size[::-1]])[0]
```

```python
for score, label, box in zip(results["scores"], results["labels"], results["boxes"]):
    if score < 0.3:
        continue
    box = [round(i, 2) for i in box.tolist()]
    print(f"Detected {label} with confidence {round(score.item(), 3)} at {box}")
```

```python

draw = ImageDraw.Draw(converted_image)

for score, label, box in zip(
    results["scores"], results["labels"], results["boxes"]
):
    if score < 0.3:
        continue

    box = [round(i, 2) for i in box.tolist()]
    x1, y1, x2, y2 = box

    draw.rectangle([x1, y1, x2, y2], outline="red", width=3)

    text = f"{label} ({round(score.item(),2)})"
    draw.text((x1, y1 - 10), text, fill="red")

converted_image.show()
```

```python
chat_loaders = {}
generator = ChatGroq(
    model_name="llama-3.1-8b-instant",
    api_key=""
)
def get_session_history(session_id: str) -> BaseChatMessageHistory:
    if session_id not in chat_loaders:
        chat_loaders[session_id] = ChatMessageHistory() 

    return chat_loaders[session_id]
from langchain.prompts import ChatPromptTemplate

gen_prompt = ChatPromptTemplate.from_messages([
    ("system", """
You are a UI action detector.

Your task is to extract texts from OCR output that likely trigger
a job application process when clicked.

IMPORTANT:
- Do NOT look for exact words.
- Instead, identify text based on MEANING (semantic intent).

A text is valid if it implies:
- starting an application
- continuing an application
- submitting an application
- submitting my resume

Examples of intent (NOT exact matches):
- anything similar to applying
- anything indicating continuation (next/continue)
- anything indicating submission

IGNORE:
- navigation (Home, About, Jobs)
- social actions (Share, Save)
- authentication (Login, Sign in)

OUTPUT RULES:
- Return ONLY a valid Python list of strings
- Each string must EXACTLY match text from OCR
- No explanations
- No extra text
- If nothing found → return []

"""),

    ("user", """
OCR Text:

Feedback:
{feedback}

Return ONLY a Python list.
""")
])
desc_prompt = ChatPromptTemplate.from_messages([
    ("system", """
You are a strict validator.

Your job is to verify whether each item in the generator's output:
1. EXISTS EXACTLY in the OCR text
2. Is relevant to job application actions

Rules:
- If any item is not present → flag it
- If any item is irrelevant → flag it
- Suggest corrections if needed

Return concise feedback only.
"""),

    ("user", """
OCR Text

Generator Output:
{response}

Check:
- Are all items present in OCR text?
- Are they valid job-application-related actions?

Return feedback.
""")
])


gen_chain = gen_prompt | generator
desc_chain = desc_prompt | generator 

gen_runner = RunnableWithMessageHistory(
    gen_chain,
    get_session_history,
    input_messages_key="feedback",      
    history_messages_key="chat_history"
)

desc_runner = RunnableWithMessageHistory(
    desc_chain,                      
    get_session_history,
    input_messages_key="response",
    history_messages_key="chat_history"
)

config_g = {"configurable": {"session_id": "gen_session"}}
config_d = {"configurable": {"session_id": "desc_session"}}

feedback = ""

for i in range(3):
    print(f"\n--- Iteration {i+1} ---")

    # Generator step
    response = gen_runner.invoke(
        {
            "feedback": feedback
        },
        config=config_g
    )

    # Discriminator step
    feedback = desc_runner.invoke(
        {
            "response": response
        },
        config=config_d
    )

    print("Generator Output:\n", response)
    print("Discriminator Feedback:\n", feedback)
```

```python
from langchain.agents import AgentExecutor, create_tool_calling_agent
from langchain_core.tools import tool
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_groq import ChatGroq
import os
from langchain_community.document_loaders import PyPDFLoader
from langchain.prompts import ChatPromptTemplate, MessagesPlaceholder
```

```python
HF_TOKEN=os.getenv("HF_TOKEN")
```

```python
embedding=HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
docs=PyPDFLoader("resume.pdf")
docs=docs.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter=RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=20)
splitted_docs=splitter.split_documents(docs)
```

```python
from langchain_community.vectorstores import FAISS

db=FAISS.from_documents(splitted_docs, embedding=embedding)
```

```python
@tool
def helper1(request:str)->str:
    """Implement the function that you want to obtain from my resume. Make sure you only pass in the keyword and based on that keyword match i will be givign you top 5 elements"""

    return db.similarity_search(request)
generator = ChatGroq(
    model_name="llama3-70b-8192",
    api_key=""
)
```

```python
tools=[helper1]
prompt=ChatPromptTemplate.from_messages([
    ("system", "You are a professional Resume alteration maker, when provided with certain requirements you update the resume according to that job requirements."),
    ("user", "Here is the job requirements: {job} and here is my resume {resume}."),
    MessagesPlaceholder("agent_scratchpad"),
    ("assistant", "Okay i understood your requirements and the resume too here are the changes")
])
agent=create_tool_calling_agent(tools=tools, prompt=prompt, llm=generator)
```

```python
agent=AgentExecutor(agent=agent, tools=tools)
```

```python
resume_text = "\n".join([doc.page_content for doc in docs])

agent.invoke({
    "job": "Machine learning engineer",
    "resume": resume_text
})
```
